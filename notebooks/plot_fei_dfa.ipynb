```python
def process_file(path):
    df = pd.read_pickle(path)
    agg = df.groupby('Group', sort=False).agg(
        ID=('ID', 'first'),
        DFA_mean=('DFA', 'mean'),
        DFA_std=('DFA', 'std'),
        fEI_mean=('fEI', 'mean'),
        fEI_std=('fEI', 'std'),
        fEI_raw_mean=('fEI_raw', 'mean'),
        fEI_raw_std=('fEI_raw', 'std'),
    ).reset_index()
    return agg


def process_all_files(paths):
    dfs = [process_file(path) for path in paths]
    return pd.concat(dfs, ignore_index=True)
```

```python
df_jext32 = process_all_files(Jext32_paths)
df_jext32
```

```python
def plot_results(df):
    features = ['DFA', 'fEI', 'fEI_raw']
    params = df['ID'].unique()
    n_params = len(params)

    fig, axes = plt.subplots(len(features), n_params, figsize=(5 * n_params, 4 * len(features)), squeeze=False)

    for col, param in enumerate(params):
        param_df = df[df['ID'] == param].copy()
        param_df['_sort_key'] = param_df['Group'].apply(
            lambda g: tuple(abs(v) for v in g) if isinstance(g, tuple) else abs(g)
        )
        param_df = param_df.sort_values('_sort_key')
        x = range(len(param_df))
        labels = [str(g) for g in param_df['Group']]

        for row, feat in enumerate(features):
            ax = axes[row, col]
            mean = param_df[f'{feat}_mean']
            std = param_df[f'{feat}_std']

            ax.plot(list(x), mean.values, marker='o')
            ax.fill_between(list(x), (mean - std).values, (mean + std).values, alpha=0.3)
            ax.set_xticks(list(x))
            ax.set_xticklabels(labels, rotation=45, ha='right', fontsize=8)

            if col == 0:
                ax.set_ylabel(feat)
            if row == 0:
                ax.set_title(param)

    fig.tight_layout()
    return fig
```

```python
fig_jext32 = plot_results(df_jext32)
```
